# Train

```python
from datasets import load_dataset

data = load_dataset("mandarjoshi/trivia_qa", "rc", split="train")

import random
import json
import pickle

save_file_name = "triviaqa_train"
sample_size = 100
seed = 42

random.seed(seed)

indices = list(range(len(data)))
random.shuffle(indices)

with open(f"{save_file_name}_s{seed}_indices.pkl", "wb") as f:
    pickle.dump(indices, f)

data = [data[i] for i in indices[:sample_size]]

with open(f"{save_file_name}_s{seed}_sample{sample_size}.json", "w") as json_file:
    json.dump(data, json_file, indent=4)
```

```python
from datasets import Dataset

from collections import defaultdict

result = defaultdict(list)


for item in data:
    for key, value in item.items():
        result[key].append(value)

result = dict(result)

dataset = Dataset.from_dict(result)
dataset.push_to_hub("alckasoc/triviaqa_expel_train_100")
```
